```python
import ee
import geemap
import pandas as pd

# =====================================================
# 1. Initialize Google Earth Engine
# =====================================================

ee.Initialize(project="geospatialfinalproject")

folder = "projects/geospatialfinalproject/assets/Hurricane_Helene_BA"

# Load your saved Earth Engine assets
inputs = ee.Image(f"{folder}/helene_21band_inputs")
labels = ee.Image(f"{folder}/helene_binary_labels")
tiles = ee.FeatureCollection(f"{folder}/helene_valid_tiles")

print("Earth Engine connected!")
print("Saved tiles:", tiles.size().getInfo())


# =====================================================
# 2. Get all 21 input band names
# =====================================================

band_names = inputs.bandNames().getInfo()

df = pd.DataFrame({
    "Band": band_names,
    "Category": [
        "Pre-Hurricane" if band.startswith("pre_")
        else "Post-Hurricane" if band.startswith("post_")
        else "Change Feature"
        for band in band_names
    ]
})

print("Total bands:", len(band_names))
display(df)


# =====================================================
# 3. Display Pre-Hurricane Sentinel-2 RGB
# =====================================================

rgb_vis = {
    "min": 0.0,
    "max": 0.12,
    "gamma": 1.2
}

Map_pre = geemap.Map()
Map_pre.centerObject(tiles, 9)

Map_pre.addLayer(
    inputs,
    {
        **rgb_vis,
        "bands": ["pre_B4", "pre_B3", "pre_B2"]
    },
    "Pre-Hurricane Sentinel-2"
)

display(Map_pre)


# =====================================================
# 4. Display Post-Hurricane Sentinel-2 RGB
# =====================================================

Map_post = geemap.Map()
Map_post.centerObject(tiles, 9)

Map_post.addLayer(
    inputs,
    {
        **rgb_vis,
        "bands": ["post_B4", "post_B3", "post_B2"]
    },
    "Post-Hurricane Sentinel-2"
)

display(Map_post)
```
